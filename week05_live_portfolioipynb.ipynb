```python
import os

os.makedirs("work/docs", exist_ok=True)

dns_text = """# How the Internet Finds a Website: A Plain-Words DNS Walkthrough
**Author:** Rafia Laeeq
**Track:** General AI Fluency | Week 5 Task 04

---

### 1. What DNS Actually Does
Computers do not understand human names like `google.com` or `rafialaeeq.netlify.app`. They only understand numeric IP addresses (like `104.198.14.52`), which point to a physical server connected to the web.

The **Domain Name System (DNS)** is the internet's phonebook. Its only job is to take human-friendly names that we can easily remember and translate them into machine-readable IP addresses so our browsers know where to fetch files.

---

### 2. The Step-by-Step Journey (From Typing a URL to Seeing the Page)

When someone types `https://rafialaeeq.netlify.app` into their browser and presses Enter, four steps happen in fractions of a second:

1. **The Resolver (The Internet's Messenger):**
   Your browser asks your local DNS resolver (usually managed by your internet service provider or public services like Cloudflare `1.1.1.1` / Google `8.8.8.8`): *"Do you know the IP address for this site?"*

2. **The Root and TLD Servers (The Phonebook Directory):**
   If the resolver doesn't already have it memorized in its cache, it checks with a Root Server. The Root Server directs it to the `.app` Top-Level Domain (TLD) registry, which directs it to Netlify's specific nameserver.

3. **The Authoritative Nameserver (The Final Answer):**
   The authoritative nameserver is the official host holding the master records for the domain. It looks up the requested address in its record list and hands back the exact destination IP address or alias to the resolver.

4. **The Response & Handshake (The Delivery):**
   The resolver sends that address back to your browser. Your browser immediately connects to Netlify's server, establishes a secure encrypted HTTPS handshake, requests `index.html`, and displays the portfolio on your screen.

---

### 3. What is a CNAME Record?
A **CNAME (Canonical Name) record** is an alias. Instead of pointing a domain name directly to a hardcoded numeric IP address (which is what an **A Record** does), a CNAME points one domain name to another domain name.

- **Example:** If I buy a custom domain like `rafialaeeq.com` in the future and want `www.rafialaeeq.com` to show my Netlify site, I create a CNAME record:
  - **Host:** `www`
  - **Type:** `CNAME`
  - **Value:** `rafialaeeq.netlify.app`
- **Why it matters:** Hosting platforms like Netlify frequently rotate their underlying IP addresses for load balancing and security. A CNAME record ensures that even if Netlify changes their servers behind the scenes, my custom domain automatically follows the alias without breaking.
"""

with open("work/docs/w05_dns_walkthrough.md", "w") as f:
    f.write(dns_text)

print("Saved DNS walkthrough to work/docs/w05_dns_walkthrough.md")
```
